### 1.2. Préparation des fonctions

```python
def get_pdf_page_total(pdf_path):
        reader = PdfReader(pdf_path)
        return len(reader.pages)
```

```python
def convert_pdf_to_jpg(input, output_folder, dpi=300, jpeg_quality=75):
    pdf_name = input.stem
    total_pages = get_pdf_page_total(input)
    n_digits = max(3, len(str(total_pages)))

    for num_page in range(1, total_pages + 1):
        images = convert_from_path(input, dpi=dpi, first_page=num_page, last_page=num_page)
        page_num = str(num_page).zfill(n_digits)
        out_path = output_folder / Path(f"{pdf_name}_{page_num}.jpg")
        images[0].save(out_path, quality=jpeg_quality)
```

## 2. Traitement du corpus

```python
for filename in image_pdf:
    filepath = input_dir / Path(filename)
    convert_pdf_to_jpg(filepath, output_dir)
```
