## 1. Préparation de l'environnement de travail
### 1.1. Outils: imports de librairies

````python
import pdfplumber
import os
from pathlib import Path
````

### 1.2. Source et espace de sauvegarde

````python
input_dir = Path("/home/bizais/Documents/Master_HN/ENCrag/Download 2026-04-28T05-38-50-283Z/Corpus_raw")
output_dir = Path("/home/bizais/Documents/Master_HN/ENCrag/TextData")
os.makedirs(output_dir, exist_ok=True)
````

````python
files = sorted(input_dir.iterdir())
print(f"{len(files)} fichier(s) trouvé(s) dans {input_dir}\n")
````

````python
not_pdf = []
image_pdf = []
````

## 2. Traitement des données
Avec sauvegarde des textes

````python
for input in files:
    if input.suffix.lower() != ".pdf":
        print(f"[NON-PDF] {input.name}")
        not_pdf.append(input.name)
        continue

    stem = input.stem  # nom sans extension

    try:
        with pdfplumber.open(input) as pdf:
            n_pages = len(pdf.pages)
            n_digits = max(3, len(str(n_pages)))
            extracted_any = False

            for i, page in enumerate(pdf.pages, start=1):
                text = page.extract_text()
                page_num = str(i).zfill(n_digits)
                out_path = output_dir/f"{stem}_page_{page_num}.txt"

                if text and text.strip():
                    extracted_any = True
                    out_path.write_text(text, encoding="utf-8")
                else: # Page vide ou image
                    out_path.write_text("", encoding="utf-8")
                
                
            if not extracted_any:
                print(f"[IMAGE-PDF] {input.name} — PDF scanné ?")
                image_pdf.append(input.name)

    except Exception as e:
        print(f"[ERREUR] {input.name} — {e}")
````

## 3. Synthèse des résultats

````python
fichiers_vides = [f for f in output_dir.iterdir() if f.stat().st_size == 0]

print("\n=== BILAN ===")
print(f"Total fichiers  : {len(files)}")
print(f"Fichiers non-PDF  : {len(not_pdf)}")
print(f"PDF image (OCR)   : {len(image_pdf)}")
print(f"Fichiers vides : {len(fichiers_vides)}")
if not_pdf:
    print("  Non-PDF :", not_pdf)
if image_pdf:
    print("  Image-PDF :", image_pdf)
````
